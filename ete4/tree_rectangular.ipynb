```python
from ete4 import Tree
from ete4.smartview import Layout, PropFace

t = Tree(
    "((Mycoplasma_genitalium:0.531161,Mycoplasma_gallisepticum:0.447071)1:0.256168,(Mycoplasma_agalactiae:0.509358,Mycoplasma_hyopneumoniae:0.539988)1:0.256168);")


def draw_node(node):
    if node.is_leaf:
        yield PropFace('name', fs_min=8, fs_max=16, position="right")


layout = Layout(name="Rectangular Layout",
                draw_node=draw_node)
t.explore(layouts=[layout])
```
